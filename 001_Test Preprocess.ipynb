```python
import os
from os import listdir
from os.path import isfile, join
import string
from random import shuffle
```

```python
# This is where we supposed to put the training/testing materal
TARGET_DIR = 'test000'
# This is where we have the documents, named by their labels
SOURCE_DIR = 'base_data_small'
# Text split multiplier
SPLIT_MULTIPLIER = 10
# Number of fake guys
FAKE_GUYS_TRAIN = 10
FAKE_GUYS_TEST = 10
```

```python
# Define the basic directories
def get_source_dir():
    return os.path.join(os.getcwd(), SOURCE_DIR)

def get_target_dir():
    return os.path.join(os.getcwd(), TARGET_DIR)

def get_target_train_dir():
    return os.path.join(os.getcwd(), TARGET_DIR, 'train')

def get_target_test_dir():
    return os.path.join(os.getcwd(), TARGET_DIR, 'test')
```

```python
# List of all files in a directory
def list_files(directory):
    return [f for f in listdir(directory) if isfile(join(directory, f))]

```

```python
# Get file name
def get_filename(dir, fname):
    return os.path.join(dir, fname)

# Read a file into a String
def read_file_to_string(filename):
    try:
        with open(filename, 'r') as file:
            data = file.read().replace('\n', ' ')
            return data
    except:
        return ""
```

```python
# CLEANSING
def cleanse(text):
    words = text.split()
    table = str.maketrans('', '', string.punctuation)
    stripped = [w.translate(table) for w in words]
    return stripped

```

```python
# Smallest split
def smallest_split(textlen, multiplier = 1):
    try:
        ssp = 1
        for i in range(20):
            ssp = ssp * (i + 1)
            if ssp >= textlen:
                return (i + 1) * multiplier
    except:
        return 0
```

```python
def split_text(text_as_list, multiplier = 1):
    txtlen = len(text_as_list)
    txtsplit = smallest_split(txtlen, multiplier)
    txtdiv = txtlen//txtsplit

    main_list = []
    act_list = []

    counter = 0
    for word in text_as_list:
        counter = counter + 1
        if counter > txtdiv:
            main_list.append(act_list)
            act_list = []
            counter = 0
        act_list.append(word)
    return main_list

            
```

```python
def flat_it(multi_list):
    try:
        retval = []
        for l1 in multi_list:
            for l2 in l1:
                retval.append(l2)
        return retval
    except:
        return []
```

```python
import random
def scrumble_up(text_list):
    shuffled = random.sample(text_list, len(text_list))
    retval = []
    for i in shuffled:
        for j in i:
            retval.append(j)
    return retval
```

```python
def list_to_string(txtlist):
    return ' '.join(txtlist)

def get_a_lot(text, num):
    cln = cleanse(text)
    retval = []
    # 1st text is the original
    retval.append(text)
    # Then go for scrumbles
    for i in range(num):
        txt = split_text(cln, SPLIT_MULTIPLIER)
        sctxt = list_to_string(scrumble_up(txt))
        retval.append(sctxt)
    return retval
```

```python
# TRY to read all files
for fname in list_files(get_source_dir()):
    # Get the file content as String
    text = read_file_to_string(get_filename(get_source_dir(), fname))
    print("ABAP name:", fname, "\n\n")

    multxt = get_a_lot(text, FAKE_GUYS_TRAIN)
    for txt in multxt:
        print(txt, "\n--------------------------------------------\n")
    
    print("\n******************************************\n")
```
