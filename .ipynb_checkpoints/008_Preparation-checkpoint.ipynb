```python
import abap as aa
```

```python
# folder with the ABAP documentation texts
SOURCE_DIR = 'base_data_small'
# maximum size of the dictionary -> the target vector dimension
DICTIONARY_SIZE = 1000
```

```python
# Get the list of file names
files, labels = aa.get_filenames(SOURCE_DIR)
```

```python
print(files)
```

```python
# Labels will be our results
print(labels)
```

```python
# Read the files, tokenize and collect the number of times different words used
word_count = {}
text_collection = []
for idx, file in enumerate(files):
    texts, word_count = aa.read_file(file, word_count)
    text_collection.append(texts)
    print(labels[idx], "-------------------------->\n")
    for txt in texts:
        for stxt in txt:
            print(stxt, "\n.................................\n")
```

```python
print(len(word_count))
```

```python
print(word_count)
```

```python
# Make a vocabulary of the most frequent used words in all documents
vocabulary = aa.get_freq_words(DICTIONARY_SIZE, word_count)
```

```python
print(vocabulary)
```

```python
# Currently text_collection has exacly so many items as labels
print(len(text_collection))
```

```python
print(len(labels))
```

```python
# But it has several dimensions:
# - per label
# - list of text lists
# - list of texts
# Now convert all to a vector
# AND make the stuff for ML:
# - train_labels = list of labels (numbers)
# - train_texts = the texts
train_labels = []
train_texts = []
for lidx, lbl in enumerate(text_collection):
    print("Label:", labels[lidx])
    # First level loop
    for txt1 in lbl:
        # Second level loop
        for txt2 in txt1:
            print("\n\n", txt2, "\n\nVECTORIZED TO:\n")
            vec = aa.get_vector(txt2, vocabulary)
            print(vec[0], "\n")
            train_labels.append(lidx)
            train_texts.append(vec[0])
```

```python
print(len(train_labels))
```

```python
print(len(train_texts))
```

```python
print(train_labels)
```

```python
print(train_texts)
```

```python
# Now, make from them a numpy tensor
import numpy as np
```

```python
train_labels_np = np.asarray(train_labels)
```

```python
train_labels_np.shape
```

```python
train_texts_np = np.asarray(train_texts)
```

```python
train_texts_np.shape
```

```python
# Build the model.
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.utils import to_categorical

model = Sequential([
  Dense(64, activation='relu', input_shape=(760,)),
  Dense(64, activation='relu'),
  Dense(10, activation='softmax'),
])
```

```python
# Compile the model.
model.compile(
  optimizer='adam',
  loss='categorical_crossentropy',
  metrics=['accuracy'],
)
```

```python
# Train the model.
model.fit(
  train_texts_np,
  to_categorical(train_labels_np),
  epochs=5,
  batch_size=32,
)
```

```python
model.evaluate(
  train_texts_np,
  to_categorical(train_labels_np)
)
```

```python
# Let's try. Vectorize a text
vector = aa.get_vector("Material master", vocabulary)
```

```python
predictions = model.predict(vector)
```

```python
print(np.argmax(predictions, axis=1))
```
